##Using StructuredTool

```python

!pip install langchain.tools
!pip install langchain-core
```

```python
from pydantic import BaseModel,Field
from langchain_core.tools import StructuredTool
```

```python
class MultiplyInputs(BaseModel):
  a:int=Field(Requried=True,description="the firsrt number to multiply")
  b:int=Field(Required=True,description="the second number to multiply")
```

```python
def multiply(a:int,b:int)->float:
  """ these funtion returns the multiplication of two numbers """
  return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply,
    name="multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInputs
)
```

```python
multiply_tool.invoke({'a':20,'b':23})
```

##Using the BaseTool Class

```python
from langchain.tools import BaseTool
```

```python
class MultiplyInput(BaseModel):
  a:int=Field(Required=True,description="the firsrt number to multiply")
  b:int=Field(Required=True,description="the second number to multiply")
```

```python
class Multiply(BaseTool):
  name:str="multiply"
  description:str="these funtion returns the product of two numbers"
  args_schema:type[BaseModel]=MultiplyInput

  def _run(self,a:int,b:int)->float:
    return a*b
```

```python
tool=Multiply()
print(tool.invoke({'a':12,'b':12}))
```

```python
print(tool.name)
print(tool.args)
print(tool.description)
```

##TOOlKIT

```python
from langchain.tools import tool
@tool
def product(a:int,b:int)->float:
  """these returns the product of two numbers"""
  return a * b

@tool
def addition(a:int,b:int)->float:
  """these returns the addition of two numbers"""
  return  a + b
```

```python
class ToolKit():
  def toolkitlist():
    return [addition,product]
```

```python
tool=ToolKit.toolkitlist()

for tool in tool:
  print(f"{tool.name} => {tool.description}")
```

```python
type(tool)
```
